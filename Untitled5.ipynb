```python
# ==========================================
# ADVANCED AUTOCORRECT TOOL
# ==========================================

# Step 1: Our Smart Dictionary with "Frequency" (How often words are used)
# The higher the number, the more popular the word is in the English language.
word_frequency = {
    "the": 5000,
    "cat": 1200,
    "car": 1500,
    "hat": 400,
    "rat": 200,
    "sat": 800,
    "mat": 300
}

# Step 2: The "Candidate Generator" (Calculates Deletions, Insertions, Swaps)
def generate_edits(word):
    letters = 'abcdefghijklmnopqrstuvwxyz'
    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]

    # 1. Deletions (removing a letter)
    deletes = [L + R[1:] for L, R in splits if R]
    # 2. Transpositions (swapping neighbor letters)
    swaps = [L + R[1] + R[0] + R[2:] for L, R in splits if len(R)>1]
    # 3. Substitutions (replacing a letter)
    replaces = [L + c + R[1:] for L, R in splits if R for c in letters]
    # 4. Insertions (adding a letter)
    inserts = [L + c + R for L, R in splits for c in letters]

    # Combine all possible variations into one massive list
    return set(deletes + swaps + replaces + inserts)

# Step 3: The Brain (Filters real words and picks the most popular one)
def advanced_autocorrect(typo_word):
    # If the word is already spelled perfectly, keep it!
    if typo_word in word_frequency:
        return typo_word

    # Generate all variations that are 1 edit away
    all_variations = generate_edits(typo_word)

    # Filter out nonsense: only keep variations that exist in our dictionary
    valid_candidates = [w for w in all_variations if w in word_frequency]

    # If we found no matches, return the original word
    if not valid_candidates:
        return typo_word

    # AI CHOICE: Pick the word with the highest frequency score!
    best_word = max(valid_candidates, key=word_frequency.get)
    return best_word

# ==========================================
# TEST TIME!
# ==========================================
# Let's type a typo: "caat". It could mean "cat", "car", or "hat".
user_input = "caat"
correction = advanced_autocorrect(user_input)

print(f"User typed: '{user_input}'")
print(f"Smart Autocorrect suggestion: '{correction}' ✨")
```

```python
# Step 4: Sentence Fixer
# This function takes a whole sentence, splits it by spaces, fixes each word, and puts it back together.
def fix_entire_sentence(sentence):
    # Break the sentence into a list of words
    words = sentence.split()
    fixed_words = []

    for word in words:
        # Clean up punctuation like periods or commas so they don't mess up the logic
        clean_word = word.lower().strip(".,!?")

        # Run our smart autocorrect on the clean word
        corrected = advanced_autocorrect(clean_word)
        fixed_words.append(corrected)

    # Glue the words back together with spaces
    return " ".join(fixed_words)

# ==========================================
# TEST DRIVING A BROKEN SENTENCE!
# ==========================================
broken_sentence = "the caat sat on the mmat"
fixed_result = fix_entire_sentence(broken_sentence)

print("❌ Original Sentence:", broken_sentence)
print("✨ Autocorrected Sentence:", fixed_result)
```

```python
# The Candidate Generator (Cleverly handles millions of combinations)
def generate_edits(word):
    letters = 'abcdefghijklmnopqrstuvwxyz'
    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]

    deletes = [L + R[1:] for L, R in splits if R]
    swaps = [L + R[1] + R[0] + R[2:] for L, R in splits if len(R)>1]
    replaces = [L + c + R[1:] for L, R in splits if R for c in letters]
    inserts = [L + c + R for L, R in splits for c in letters]

    return set(deletes + swaps + replaces + inserts)

# The Brain
def advanced_autocorrect(typo_word):
    typo_word = typo_word.lower()
    if typo_word in word_frequency:
        return typo_word

    all_variations = generate_edits(typo_word)
    valid_candidates = [w for w in all_variations if w in word_frequency]

    if not valid_candidates:
        return typo_word

    # Picks the best real word candidate
    return max(valid_candidates, key=word_frequency.get)

# The Sentence Fixer
def fix_entire_sentence(sentence):
    words_list = sentence.split()
    fixed_words = []
    for word in words_list:
        clean_word = word.lower().strip(".,!?")
        corrected = advanced_autocorrect(clean_word)
        fixed_words.append(corrected)
    return " ".join(fixed_words)

# ==========================================
# TEST YOUR UNLIMITED POWER!
# ==========================================
# You can change this sentence to absolutely ANYTHING you want now!
my_test = "i want to eat an appel and fly a kitte"
result = fix_entire_sentence(my_test)

print("❌ Input:", my_test)
print("✨ Fixed:", result)
```

```python
%%writefile app.py
import streamlit as st
import nltk
from collections import Counter

# --- BACKGROUND ENGINE SETUP ---
@st.cache_resource
def load_vocab():
    nltk.download('words', quiet=True)
    from nltk.corpus import words
    return Counter(word.lower() for word in words.words())

word_frequency = load_vocab()

def generate_edits(word):
    letters = 'abcdefghijklmnopqrstuvwxyz'
    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]
    deletes = [L + R[1:] for L, R in splits if R]
    swaps = [L + R[1] + R[0] + R[2:] for L, R in splits if len(R)>1]
    replaces = [L + c + R[1:] for L, R in splits if R for c in letters]
    inserts = [L + c + R for L, R in splits for c in letters]
    return set(deletes + swaps + replaces + inserts)

def advanced_autocorrect(typo_word):
    typo_word = typo_word.lower()
    if typo_word in word_frequency:
        return typo_word
    all_variations = generate_edits(typo_word)
    valid_candidates = [w for w in all_variations if w in word_frequency]
    if not valid_candidates:
        return typo_word
    return max(valid_candidates, key=word_frequency.get)

def fix_entire_sentence(sentence):
    words_list = sentence.split()
    fixed_words = []
    for word in words_list:
        clean_word = word.lower().strip(".,!?")
        corrected = advanced_autocorrect(clean_word)
        fixed_words.append(corrected)
    return " ".join(fixed_words)

# --- MODERN WEB UI DESIGN ---
st.set_page_config(page_title="Smart AutoCorrect", page_icon="✨", layout="centered")

st.title("✨ AI-Powered Autocorrect Engine")
st.markdown("Improve text accuracy and fluency instantly. Type your paragraph below to eliminate errors.")

# Create a clean text area input
user_input = st.text_area("Your Input Text", value="i want to eat an appel and fly a kitte", height=150)

if st.button("Fix Text & Fluency", type="primary"):
    if user_input.strip():
        with st.spinner("Processing text pipeline..."):
            corrected_result = fix_entire_sentence(user_input)

        # Displaying side-by-side comparison using modern UI columns
        col1, col2 = st.columns(2)
        with col1:
            st.error("Original Text")
            st.caption(user_input)
        with col2:
            st.success("Corrected Text")
            st.write(corrected_result)
    else:
        st.warning("Please enter some text to correct.")
```
